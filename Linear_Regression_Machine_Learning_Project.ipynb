```python
#we will delet them
df_cleaned = df.dropna(subset=["gpu_type"])
```

```python
features=[
    "cpu_core_count",
        "cpu_thread_count",
            "ram_gb",
             "storage_gb",
             "gpu_type",
             "gpu_model",

]
x = df_cleaned[features]
y = df_cleaned['price']
```

```python
x.head()
```

```python
y.head()
```

```python
print("missing values in x")
print(x.isnull().sum())
print("\nTotal missing values in ", x.isnull().sum().sum())
```

```python
#2 features one numbers one string
num_features = [
     "cpu_core_count",
        "cpu_thread_count",
            "ram_gb",
             "storage_gb"
]
categorical_freatures = [
     "gpu_type",
             "gpu_model"
]
```

```python
#convert the data into numbers (one hot encoding)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LinearRegression

preprocessor = ColumnTransformer(
    transformers=[
        ('categorical',
         OneHotEncoder(handle_unknown="ignore"),
         categorical_freatures
         )
    ],
    remainder="passthrough"

)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
print("training freatures",x_train.shape)
print("training freatures",x_test.shape)

print("training freatures",y_train.shape)
print("training freatures",y_test.shape)

```

```python
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", LinearRegression())
    ]

)
```

```python
model.fit(x_train, y_train)
print("training completed")
```

```python
y_predict=model.predict(x_test)
```
